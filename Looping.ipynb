```python
#1 Print numbers from 1 to 20 using a for loop.
for i in range(1,21):
  print(i)
```

```python
#2 Print numbers from 20 to 1 using a for loop.
for i in range(20,0,-1):
  print(i)
```

```python
#3 Print all even numbers between 1 and 50
for i in range(1,51):
  if(i%2==0):
    print(i)
```

```python
#4 Print all odd numbers between 1 and 50.
for i in range(1,51):
  if(i%2!=0):
    print(i)

```

```python
#5 Print the multiplication table of 7
for i in range(1,11):
  print(f"7 x {i} = {7*i}")
```

```python
#6 Calculate the sum of numbers from 1 to 100
sum=0
for i in range(1,101):
  sum=sum+i
print(sum)
```

```python
#8 Count the number of elements in a list using a loop
name=["Nilal","Ellappan","Shamri"]
count=0
for i in name:
  count=count+1
print(count)
```

```python
# while
# 9 Print numbers from 1 to 10 using a while loop
i=1
while i<=10:
  print(i)
  i=i+1

```

```python
#10 Find the sum of even numbers from 1 to 50 using while.
total_sum = 0
i = 1
while i <= 50:
  if i % 2 == 0:
    total_sum = total_sum + i
  i = i + 1
print(total_sum)
```

```python
#11 Reverse a number using a while loop
n = 123
reversed_num = 0

while n > 0:
    digit = n % 10
    reversed_num = (reversed_num * 10) + digit
    n = n // 10

print(reversed_num)
```

```python
#12 Count the digits in a given number
n = 12345
count = 0

while n > 0:
    n = n // 10
    count = count + 1

print(count)
```

```python
#13 Ask the user to enter a number until they enter 0
while True:
    num = int(input("Enter a number (0 to exit): "))
    if num == 0:
        print("Loop terminated.")
        break
```

```python
#14 Print numbers from 1 to 20 but skip 10 using continue
for i in range(1, 21):
  if i == 10:
    continue
  print(i)
```

```python
#15 Stop the loop when the number reaches 15 using break
for i in range(1, 20):
  if i == 15:
    break
  print(i)
```
