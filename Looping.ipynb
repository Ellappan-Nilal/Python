```python
#1 Print numbers from 1 to 20 using a for loop.
for i in range(1,21):
  print(i)
```

```python
#2 Print numbers from 20 to 1 using a for loop.
for i in range(20,0,-1):
  print(i)
```

```python
#3 Print all even numbers between 1 and 50
for i in range(1,51):
  if(i%2==0):
    print(i)
```

```python
#4 Print all odd numbers between 1 and 50.
for i in range(1,51):
  if(i%2!=0):
    print(i)

```

```python
#5 Print the multiplication table of 7
for i in range(1,11):
  print(f"7 x {i} = {7*i}")
```

```python
#6 Calculate the sum of numbers from 1 to 100
sum=0
for i in range(1,101):
  sum=sum+i
print(sum)
```

```python
#7 Print each character of a given string
name="Progammming"
for i in name:
  print(i)
```

```python
#8 Count the number of elements in a list using a loop
name=["Nilal","Ellappan","Shamri"]
count=0
for i in name:
  count=count+1
print(count)
```
