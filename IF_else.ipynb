```python
#10 Create a Simple Calculator using operators (+, -, *, /)
n1 = float(input("Enter number 1: "))
op = input("Enter operator (+, -, *, /): ")
n2 = float(input("Enter number 2: "))

if op == "+":
    print("Result:", n1 + n2)
elif op == "-":
    print("Result:", n1 - n2)
elif op == "*":
    print("Result:", n1 * n2)
elif op == "/":
    if n2 != 0:
        print("Result:", n1 / n2)
    else:
        print("Error! Division by zero is not allowed.")
else:
    print("Invalid operator")
```

```python
#12 ATM Withdrawal System with PIN and Balance Verification
pin = input("Enter your 4-digit PIN: ")
correct_pin = "1234"
balance = 5000

if pin == correct_pin:
    amount = int(input("Enter the amount to withdraw: "))
    if amount <= balance:
        balance -= amount
        print("Withdrawal successful! Remaining balance:", balance)
    else:
        print("Insufficient balance.")
else:
    print("Incorrect PIN. Transaction cancelled.")
```

```python
#14 Shopping Discount Calculator based on Purchase Amount and Membership
amount = float(input("Enter your purchase amount: "))
if amount >= 1000:
    membership = input("Do you have a premium membership? (yes/no): ").strip().lower()
    if membership == "yes":
        discount = 0.20 # 20% discount
    else:
        discount = 0.10 # 10% discount
else:
    membership = input("Do you have a premium membership? (yes/no): ").strip().lower()
    if membership == "yes":
        discount = 0.05 # 5% discount
    else:
        discount = 0.0  # No discount

final_price = amount * (1 - discount)
print(f"Applied Discount: {discount * 100}%")
print(f"Final Amount to Pay: {final_price}")
```

```python
#16 Create a Basic Calculator using Match-Case
n1 = float(input("Enter number 1: "))
op = input("Enter operator (+, -, *, /): ")
n2 = float(input("Enter number 2: "))

match op:
    case "+":
        print("Result:", n1 + n2)
    case "-":
        print("Result:", n1 - n2)
    case "*":
        print("Result:", n1 * n2)
    case "/":
        if n2 != 0:
            print("Result:", n1 / n2)
        else:
            print("Error! Division by zero is not allowed.")
    case _:
        print("Invalid operator")
```

```python
#17. Traffic Signal System (Red, Yellow, Green).
signal=input("Enter the signal : ")
match signal:
  case "Red":
    print("Stop")
  case "Yellow":
    print("Get Ready")
  case "Green":
    print("Go")
  case _:
    print("Invalid signal")
```

```python
#18 Create a Food Ordering Menu
print("--- Menu ---")
print("1. Pizza\n2. Burger\n3. Sandwich\n4. Pasta")
choice = input("Enter the food item or number: ").strip().capitalize()

match choice:
    case "Pizza" | "1":
        print("You ordered Pizza. That will be Rs:10.")
    case "Burger" | "2":
        print("You ordered a Burger. That will be Rs:6.")
    case "Sandwich" | "3":
        print("You ordered a Sandwich. That will be Rs:5.")
    case "Pasta" | "4":
        print("You ordered Pasta. That will be Rs:8.")
    case _:
        print("Sorry, that item is not on the menu!")
```
