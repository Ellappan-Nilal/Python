```python
 #9 Create an Electricity Bill Calculator
unit=int(input("Enter the unit : "))
if unit<=100:
  bill=unit*5
elif unit<=200:
  bill=100*5+(unit-100)*10
else:
  bill=100*5+100*10+(unit-200)*15
print("Bill amount : ",bill)
```

```python
#10 Create a Simple Calculator using operators (+, -, *, /)
n1 = float(input("Enter number 1: "))
op = input("Enter operator (+, -, *, /): ")
n2 = float(input("Enter number 2: "))

if op == "+":
    print("Result:", n1 + n2)
elif op == "-":
    print("Result:", n1 - n2)
elif op == "*":
    print("Result:", n1 * n2)
elif op == "/":
    if n2 != 0:
        print("Result:", n1 / n2)
    else:
        print("Error! Division by zero is not allowed.")
else:
    print("Invalid operator")
```

```python
#11 Username and Password Verification (Nested If)
username = input("Enter username: ")
if username == "admin":
    password = input("Enter password: ")
    if password == "secure123":
        print("Access Granted!")
    else:
        print("Access Denied: Incorrect Password.")
else:
    print("Access Denied: Username not found.")
```

```python
#12 ATM Withdrawal System with PIN and Balance Verification
pin = input("Enter your 4-digit PIN: ")
correct_pin = "1234"
balance = 5000

if pin == correct_pin:
    amount = int(input("Enter the amount to withdraw: "))
    if amount <= balance:
        balance -= amount
        print("Withdrawal successful! Remaining balance:", balance)
    else:
        print("Insufficient balance.")
else:
    print("Incorrect PIN. Transaction cancelled.")
```

```python
#13 College Admission Eligibility based on Age and Marks
age = int(input("Enter your age: "))
if age >= 17:
    marks = float(input("Enter your marks percentage: "))
    if marks >= 60.0:
        print("Admission Eligible!")
    else:
        print("Not Eligible: Marks must be 60% or higher.")
else:
    print("Not Eligible: Age must be 17 or older.")
```

```python
#14 Shopping Discount Calculator based on Purchase Amount and Membership
amount = float(input("Enter your purchase amount: "))
if amount >= 1000:
    membership = input("Do you have a premium membership? (yes/no): ").strip().lower()
    if membership == "yes":
        discount = 0.20 # 20% discount
    else:
        discount = 0.10 # 10% discount
else:
    membership = input("Do you have a premium membership? (yes/no): ").strip().lower()
    if membership == "yes":
        discount = 0.05 # 5% discount
    else:
        discount = 0.0  # No discount

final_price = amount * (1 - discount)
print(f"Applied Discount: {discount * 100}%")
print(f"Final Amount to Pay: {final_price}")
```
